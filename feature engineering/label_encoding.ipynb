### label Encoding
#### label encoding and ordinal encoding are two techniques usedd to encode categorical data as numerical data 

#### Label encoding involves assigning a unique numerical label to each category in the variable. The labels are usually assigned in alphabetical order or based on the frequency of the categories. for exxample, if we have a categorical variable "color" with three possible values (red,green,blue), we can represent it using label encoding as follows:

#### 1.Red:1
#### 2.Green:2
#### 3.BLue:3

```python
import pandas as pd
from  sklearn.preprocessing import OneHotEncoder 
```

```python
df = pd.DataFrame({
    'color' : ['red', 'blue', 'green', 'green', 'red', 'blue']
})
```

```python
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder
lbl_encoder = LabelEncoder()
```

```python
lbl_encoder.fit_transform(df['color'])
```

```python
lbl_encoder.transform(['green'])
```

#### Ordinal Encoding
##### Ordinal Encoding is used to convert categorical data that has a meaningful order/ranking into numbers.

##### Example 
###### 1. Small  → 0
###### 2.Medium → 1
###### 3.Large  → 2

```python
### Ordinal encoding

from sklearn.preprocessing import OrdinalEncoder
```

```python
### Create a sample dataframe with an ordinal variable 

df = pd.DataFrame({
    'size': ['small','medium','large','medium','small','large']
})
```

```python
df
```

```python
## create an instance of ordinal encoder and then fit_tranforms

encoder = OrdinalEncoder(categories = [['small','medium','large']]) 
```

```python
encoder.fit_transform(df[['size']])
```

```python
encoder.transform([['small']])
```
