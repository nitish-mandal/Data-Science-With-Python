##### SMOTE stands for Synthetic Minority Over-sampling Technique.
##### It is used when your dataset is imbalanced, meaning one class has many more samples than another class.

##### Example
##### Suppose you have a dataset for fraud detection:
##### Class	Number of records
##### Normal transaction	950
##### Fraud transaction	50
##### Here, Fraud is the minority class.
##### A machine-learning model may become biased toward the majority class because it sees 950 normal examples but only 50 fraud examples.

##### SMOTE helps by creating new synthetic samples for the minority class.

```python
from sklearn.datasets import make_classification
```

```python
x, y = make_classification(n_samples = 1000, n_redundant=0, n_features=2, n_clusters_per_class=1, weights=[0.90], random_state = 12)
```

```python
import pandas as pd
df1 = pd.DataFrame(x,columns=['f1','f2'])
df2 = pd.DataFrame(y,columns=['target'])
final_df = pd.concat([df1,df2],axis=1)
final_df.head()
```

```python
final_df['target'].value_counts()
```

```python
import matplotlib.pyplot as plt 
plt.scatter(final_df['f1'],final_df['f2'],c=final_df['target']) 
```

```python
from imblearn.over_sampling import SMOTE
```

```python
## transform the dataset 
oversample = SMOTE()
X,y = oversample.fit_resample(final_df[['f1','f2']], final_df['target'])
```
