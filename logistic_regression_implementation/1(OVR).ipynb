```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns 
%matplotlib inline 
```

```python
from sklearn.datasets import make_classification
```

```python
### create the dataset
X,y = make_classification(n_samples=1000,n_features=10,n_classes=2,random_state=42)
```

```python
pd.DataFrame(X)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.30,random_state=42)
```

```python
## model training 
from sklearn.linear_model import LogisticRegression
logistic = LogisticRegression()
```

```python
logistic.fit(X_train,y_train)
```

```python
## prediction for test dataset 
y_pred = logistic.predict(X_test)
print(y_pred)
```

```python
logistic.predict_proba(X_test)
```

```python
## performance metrics
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

```python
score=accuracy_score(y_test,y_pred)
print(score)
cm = confusion_matrix(y_test,y_pred)
print(cm)
print(classification_report(y_test,y_pred))
```

#### Grid Search CV

| Method | Purpose |
|---|---|
| `randomcv.fit()` | Performs tuning and trains the best model |
| `randomcv.best_score_` | Best average CV accuracy |
| `randomcv.best_params_` | Best hyperparameters |
| `randomcv.predict()` | Predicts test data |
| `accuracy_score()` | Calculates accuracy |
| `classification_report()` | Shows precision, recall and F1-score |
| `confusion_matrix()` | Shows correct and incorrect classifications |

```python
from sklearn.model_selection import StratifiedKFold
cv = StratifiedKFold()
```

```python
from sklearn.model_selection import GridSearchCV
grid = GridSearchCV(estimator=model,param_grid=param,scoring = 'accuracy',cv=cv, n_jobs= -1 )
```

```python
grid
```

```python
grid.best_params_
```

```python
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
y_pred = grid.predict(X_test)
```

```python
score = accuracy_score(y_pred,y_test)
print(score)
print(classification_report(y_pred,y_test))
print(confusion_matrix(y_pred,y_test))
```

#### Logistic Regression for Imbalanced Dataset

```python
## Generate and plot a synthetic imbalaced classification dataset
from collections import Counter
from sklearn.datasets import make_classification
```

```python
## Imbalanced dataset
X,y = make_classification(n_samples=10000,n_features=2,n_clusters_per_class=1,n_redundant=0,
                          weights=[0.99],random_state=10)
```

```python
X
```

```python
Counter(y)
```

```python
import seaborn as sns
```

```python
import pandas as pd
sns.scatterplot(x=X[:, 0], y=X[:, 1], hue=y)
plt.show()
```

```python
### split the dataset into train and test
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.25,random_state=0)
```

```python
### Hyperparameter tunning 
from sklearn.linear_model import LogisticRegression
model =  LogisticRegression()
## Penalty is a regularization technique that helps control model complexity and reduce overfitting
penalty = ['l1','l2','elasticnet']
## What is C?

#C controls the strength of regularization.

#It works inversely:
                     # {Regularization Strength} = 1/C

#C Value	    Regularization
#100      	    Very weak
#10      	    Weak
#1.0	        Moderate
#0.1	        Strong
#0.01    	    Very strong

#Remember:
# - Small C → Strong regularization.
#- Large C → Weak regularization.
#We give multiple values because we don't know which one will perform best on our dataset.
c_values = [100,10,1.0,0.1,0.01]
## the solver helps Logistic Regression learn the best weights from training data.
solver = ['newton-cg','lbfgs','liblinear','sag','saga']
class_weight = [{0:w,1:y} for w in [1,10,50,100] for y in [1,10,50,100]]
```

```python
# Combine Hyperparameters into a Dictionary
ams = dict(penalty=penalty,C=c_values,solver=solver,class_weight=class_weight)
```

```python
param
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import StratifiedKFold
cv = StratifiedKFold()
grid  = GridSearchCV(estimator=model,param_grid=param,scoring ='accuracy',cv=cv)
```

```python
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
y_pred = grid.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report,confusion_matrix
```

```python
score = accuracy_score(y_pred,y_test)
print(score)
classification_report(y_pred,y_test)
confusion_matrix(y_pred,y_test)
```
