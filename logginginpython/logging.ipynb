```python
 ## logging is  a crucial aspect of any application, providing a way track event, errors, and operational operation
 #  information. python's built in logging modules offers  flexible framework for emitting log messages from python
 #  programs. In this lesson, we will cover the basics of logging, including how to configure logging, log levels, 
 # and the best practices for using logging in python applications. 
```

```python
## it is used to record messages about what a program is doing while it runs - for debugging, monitoring and error tracking.
```

```python
import logging 

## configure the basic logging settings
logging.basicConfig(level=logging.DEBUG)

## log messages with the different severnity levels
logging.debug("This is a debug message")
logging.info("this is an info messsage ")
logging.warning("this is a warning messsage ")
logging.error("This is an error message ")
logging.critical("This is a critical message ")
```

### log levels
##### DEBUG: Details information , typically, of interest only when diagnosing problems.
##### INFO: Confirmation that things are working as expected
##### WARNING: An indication that something unexpected happend or indicative of some problen in the near future (e.g., 'disk space low'). the software is still working as expected 
##### ERROR: due to more serious problem, the software has not been able to perform some function
##### CRITICAL: a very serious error, indicating that the program itself may be unable to continue working 

```python
## configuring logging 
import logging 


logging.basicConfig(
    filename='app.log',
    filemode='w',
    level=logging.DEBUG,
    format= '%(asctime)s- %(name)s-%(levelname)s-%(message)s',
    datefmt= '%Y-%M-%d %H:%M:%S'
)

## log messages with the different severnity levels
logging.debug("This is a debug message")
logging.info("this is an info messsage ")
logging.warning("this is a warning messsage ")
logging.error("This is an error message ")
logging.critical("This is a critical message ")
```
