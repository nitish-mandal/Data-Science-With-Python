##### Memory management in Python is the process of storing data in memory and automatically removing unused data. Python does this using reference counting, garbage collection, and internal optimizations. These techniques help Python use memory efficiently and allow developers to create faster, more reliable programs.

##### 3. Reference Counting is a memory management technique in Python that counts how many variables (references) are pointing to an object.
If the reference count is greater than 0, the object stays in memory.
If the reference count becomes 0, Python knows the object is no longer needed and can remove it from memory.

```python
import sys

a = []
## 2 ( one refrence from 'a' and one from getrefcount())
print(sys.getrefcount(a))
```

```python
b=a
print(sys.getrefcount(a))
```

```python
del b   ## can be delete reference b 
print(sys.getrefcount(a))
```

### garbage colection 
Garbage Collection (GC) is a memory management process in Python that automatically finds and removes objects that are no longer being used, so their memory can be freed and reused.

```python
import gc 
## enable garbage collection 
gc.enable()
```

```python
gc.disable()
```

```python
## Normally, Python runs the garbage collector automatically. But if you want to clean memory immediately, you can use gc.collect()

gc.collect()
```

```python
## gc.get_stats() is a function in Python that shows statistics (information) about the Garbage Collector, such as how many objects it has collected and how many collection cycles have run
print(gc.get_stats())
```

```python
## get unreable objects
# An unreachable object is an object that cannot be accessed by any variable or reference in the program.
# Since no part of the program can use it anymore, it becomes garbage, and Python's Garbage Collector (GC) can remove i
print(gc.garbage)
```

#### memory management best practices

1. Local variables are variables created inside a function. They exist only while the function is running
After the function finishes, Python automatically removes them from memory.

2. A circular reference happens when two or more objects refer (point) to each other

3. Use Generators: Generators produce items one at a time and only keep one item in memory at a time, making them memory efficient.

4. Explicitly Delete Objects: Use the del statement to delete variables and objects explicitly.

5. Profile Memory Usage: Memory profiling means checking how much memory your program is using. Use memory profiling tools like tracemalloc and memory_profiler to identify memory leaks and optimize memory usage.

```python
## Handling circular reference 
import gc 

class MyObject :
    def __init__(self,name):
        self.name=name
        print(f"Object {self.name} created ")

    def __del__(self):
        print(f"Object {self.name} deleted") 

## create circular refrence 
obj1 = MyObject("obj1")
obj2 = MyObject("obj2") 
obj1.ref = obj2
obj2.ref = obj1 

del obj1
del obj2

## manually trigger the garbage collection 
gc.collect()
```

```python
## Generator for memory efficiency

def generate_numbers(n):
    for i in range(n):
        yield i    # Normally, a function returns all results at once using return. #  With yield, the function returns one value at a time and pauses after each value   #This saves memory

## using this generator 
for num in generate_numbers(100000) :   
    print(num)
    if num>10:
        break               
                   
```

```python
from tracemalloc import Snapshot
import tracemalloc 

## profiling memory 

def create_list():
    return [i for i in range(10000)]

def main():
    tracemalloc.start()

    create_list()

    Snapshot = tracemalloc.take_snapshot()
    top_stats = Snapshot.statistics('lineno')

    print("[ Top 10 ]")
    for stat in top_stats[::]:
        print(stat)

main()        
```
