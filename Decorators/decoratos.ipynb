### Decorators

Decorstors are a powerful and flexible feature in python that allows you to modify the behaviour of a function or class method . they are commonly used to add functionality to functios or method without modifying their actual code. this lesson covers the basics of decorators , including how to create and use them

```python
## Decorator..
def main_welcome(func):
   
    def sub_welcome_method():
        print("welcome to the advanced python course")
        func()
        print("please learn this conceot properly ")
    return sub_welcome_method()
```

```python
def coure_introduction():
    print("this is na advanced python course")

coure_introduction()    
```

```python
main_welcome(coure_introduction)  ## without this we cancall the function ...  using @main_welcome
```

```python
@main_welcome
def coure_introduction():
    print("this is an advanced python course")
```

```python
## Decorator..
def my_decorator(func):
   
    def wrapper():
        print("Something happening before the function is callede..")
        func()
        print("Something happening before the function is callede.. ")
    return wrapper
```

```python
@my_decorator
def say_hello():
    print("hello!")
```

```python
say_hello()
```

```python
## Decorators with arguments

def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator        
```

```python
@repeat(3)
def say_hello():
    print("hello")
```

```python
say_hello()
```
