iterators are advanced python concepts that allow for efficient looping and memory management iterators provide a way access elements of a collection sequentially without exposing the underline structure 

```python
my_list = [1,2,3,4,5,6,7]
for i in my_list:
    print(i)
```

```python
type(my_list)
```

```python
print(my_list)
```

```python
## Iterator 
iterator = iter(my_list)
print(type(iterator))
```

```python
iterator
```

```python
## Iterate through all the element 
next(iterator)
```

```python
## to handle this exception we use try,catch
iterator = iter(my_list)
```

```python

try :
    print(next(iterator))
except   StopIteration:
    print("There are no element in the iterator ")  
```
