### Pandas - DataFrame And Series

pandas is a powerful data manipulation library in python , widely used for data analysis and data cleaning. it provides two primary data structures: Series is a one-dimensional array-like object, while a DataFrame is a two-dimensional, size-mutable, and potentially heterogeneous tabular data structure with labeled axes(rows and columns)

```python
import pandas as pd
```

```python
##Series
## A pandas Series is a one-dimensional array-like object that can hold any data type. It is similar to column in a table 
## it is default...
import pandas as pd
data = [1,2,3,4,5]
series = pd.Series(data)
print("Series /n", series)
```

```python
## Create a Series From dictionary
data = {'a':1, 'b':2, 'c':3}
series_dict = pd.Series(data)
print(series_dict)
```

```python
## we can give different index according to requirement..

data = [10,20,30]
index = ['A','B','C']
pd.Series(data,index= index)
```

```python
## Dataframe
## create a dataframe from a dictionary of list

data = {
    'Name':['Nitish','Priyansu','Aniket'],
    'Age':[20,21,19],
    'City':['Banglore','New York','Florida']
}
df= pd.DataFrame(data)
print(df)
print(type(df))
```

```python
## Create a Data frame from a list of dictionaries..

data=[
    {'Name':'Nitish','Age':20,'City':'Dehradun'},
    {'Name':'Priyanshu','Age':21,'City':'Dehradun'},
    {'Name':'Aniket','Age':19,'City':'Dehradun'},
    {'Name':'Dipesh','Age':20,'City':'Dehradun'}

]
df = pd.DataFrame(data)
print(df)
print(type(df))
```
