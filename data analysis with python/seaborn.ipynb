#### Data Visualization with seaborn

seaborn is a python visualization library based on Matplotlib that provides a high-level interface for drawing attractive and informative statistical graphics. Seaborn helps in creating complex visualization with just a few lines of codes. in this lesson, we will cover the basics of seaborn, including creating various types of plots and customizing them.. 

```python
import seaborn as sns
```

```python
### Basic Plotting With Seaborn
tips = sns.load_dataset('tips')
tips
```

```python
## create a scatter plot
import matplotlib.pyplot as plt

sns.scatterplot(x= 'total_bill', y= 'tip',data=tips)
plt.title("Scatter plot of total Bill vs tips")
plt.show()
```

```python
sns.lineplot(x='size', y='total_bill',data=tips)
plt.title("Line Plot of total bill by size")
plt.show()
```

```python
## Categorical plots
## Bar plot
sns.barplot(x='day',y='total_bill',data=tips)
plt.title('Bar plot of total Bill by day')
plt.show()

```

```python
## Box plot
sns.boxplot(x="day", y="total_bill",data=tips)
```

```python
## violin plot

sns.violinplot(x='day', y='total_bill', data=tips)
```

```python
## Histograms
sns.histplot(tips['total_bill'],bins=10,kde=True)
 ##Bins = jitne groups me data divide hoga.
 ## KDE = smooth curve (continuous probability)
```

```python
## KDE plot
sns.kdeplot(tips['total_bill'],fill=True)
```

```python
## pairplot
sns.pairplot(tips)
```

```python
## Heatmap
## relation btwn one feature to another..

corr= tips[['total_bill','tip','size']].corr()
corr
```

```python
sns.heatmap(corr,annot=True,cmap='coolwarm')
```

```python
import pandas as pd 
sales_df = pd.read_csv('sales_data.csv')
sales_df.head()
```

```python
## Plot total  quantity
plt.Figure(figsize=(10,6))
sns.barplot(x='Quantity',y='Total Amount',data=sales_df,estimator=sum)
plt.title('Total quantity')
plt.xlabel('product')
plt.ylabel('Total Amount')
plt.show()
```

```python
## Plot total sales by region 
plt.Figure(figsize=(10,6))
sns.barplot(x='Product Category',y='Total Amount',data=sales_df,estimator=sum)
plt.title('Total sales by Region')
plt.xlabel('region')
plt.ylabel('Total Amount')
plt.show()
```
