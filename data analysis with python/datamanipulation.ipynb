```python
## Handling missing value
df.isnull().sum()
```

```python
df_filled=df.fillna(0)
```

```python
## Filling missing values with the mean of the column
df['Numb_doors']=df['Doors'].fillna(df['Doors'].mode())
df
```

```python
## if want to change data type 
df.dtypes
```

```python
## Renaming columns
df= df.rename(columns={'Doors':'numb_doors'})  ## keyvalue pairs
df.head()
```

```python
# change datatypes
## fill NaN values with the column mode and convert to string
df['numb_doors'] = df['numb_doors'].fillna(df['numb_doors'].mode()[0]).astype(str)  ## or df['Mix_colour']=df['Colour'].fillna(df['Colour'].mode()[0]).astype(str)
df.head()
```

```python
df
```

```python
## Data aggregating And Grouping
df.head()
```

```python
grouped_mean = df.groupby('Colour')['Odometer (KM)'].mean()
print(grouped_mean)
```

```python
## if want to groupby with two categorial variables
grouped_mean = df.groupby(['Make','Colour'])['Price'].sum()
print(grouped_mean)
```

```python
## we can also calculate mean
df.groupby(['Make','Colour'])['Price'].mean()
```

```python
## Aggregate multiple Functions
grouped_agg = df.groupby(['Make'])['Price'].agg(['mean','sum','count'])
print(grouped_agg)
```

```python
## Merging And Joining Dataframes
# create sample Dataframes

df1= pd.DataFrame({'Key': ['A','B','C'], 'Value': [1,2,3]})
df2= pd.DataFrame({'Key': ['A','B','D'], 'Value': [4,5,6]})
```

```python
df1
```

```python
df2
```

```python
## Merge Dataframe on the 'Key Columns'
pd.merge(df1,df2,on="Key",how="inner")  ## on means which column and how means how should happen
```

```python
pd.merge(df1,df2,on="Key",how="outer")
```

```python
pd.merge(df1,df2,on="Key",how="left")
# Left Merge = All rows from df1 + matching rows from df2
#No match → NaN
```

```python
pd.merge(df1,df2,on="Key",how="right")
# Keep all rows from the RIGHT DataFrame (df2)
#👉 Match rows from the left DataFrame (df1)
#👉 If a key from df2 is NOT found in df1 → fill with NaN
```
