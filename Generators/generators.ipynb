Generators are a simpler way to create iterators. They use the yield  keyword to produce a series of values lazily, which means they generte values on the fly and do not store them in memory 

```python
def square(n):
    for i in range(3):
        yield i**2
square(3)    
```

```python
for i in range(3):
    print(i)
```

```python
a = square(3)
a
```

```python
next(a)
```

```python
def my_generator():
    yield 1
    yield 2
    yield 3

  
```

```python
gen = my_generator()
gen  
```

```python
next(gen)
```

```python
for val in gen:
    print(val)
```
